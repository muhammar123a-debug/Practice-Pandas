```python
import pandas as pd
```

```python
df = pd.read_csv('Datasets/amazon_product.csv')
```

```python
"""
Deleting redundent columns
Renaming columns name
Dropping dupplicates
Remove NAN value in our datasets
Check the some more Transform
"""
```

```python
df.columns
```

```python
df.drop(columns=['Unnamed: 0', 'asin','product_url', 'product_photo',
       'product_num_offers', 'product_minimum_offer_price','is_amazon_choice', 'is_prime', 'climate_pledge_friendly',
       'sales_volume', 'delivery', 'has_variations',], inplace=True)
```

```python
df.columns.str.capitalize().str.strip()
```

```python
df.duplicated().sum()
```

```python
df['is_best_seller'] = df['is_best_seller'].apply(lambda x: 1 if x == True else 0)
```

```python
df['product_price'] = df['product_price'].str.replace('$','').astype(float)
```

```python
df['product_original_price'] = df['product_original_price'].str.replace('$','')
```

```python
df['product_price']
```

```python
df.drop(columns=['product_availability','unit_price','unit_count'], inplace=True)
```

```python
df['product_original_price'] = df['product_original_price'].fillna(0)
```

```python
df['product_num_ratings'] = df['product_num_ratings'].fillna(df['product_num_ratings'].median())
```

```python
df['product_star_rating'] = df['product_star_rating'].fillna(df['product_star_rating'].median())
```
